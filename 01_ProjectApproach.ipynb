## 5. Project Workflow & Implementation Approach

### Phase 1: Problem Understanding & Dataset Ingestion
- Define project scope focusing on Indian banking transaction records and UPI payment patterns.
- Review dataset schema, attribute constraints, and data dictionaries.
- Establish core analytical targets aligning with data transformation and visualization objectives.

### Phase 2: Data Cleaning & Transformation (Pandas)
- Load the raw CSV records into Pandas DataFrames.
- Profile column types, identify schema inconsistencies, and convert temporal columns to appropriate datetime formats.
- Drop non-informative or redundant identifier columns.
- Audit missing values and perform mean imputation on continuous numerical columns.
- Derive engineered features (e.g., transaction hour, day of week) for downstream analysis.

### Phase 3: Numerical Operations & Statistical Metrics (NumPy)
- Convert key numerical features (transaction amounts, account balances) into NumPy ndarrays.
- Calculate summary statistics including mean, standard deviation, and variance using vectorized operations.
- Determine quartile thresholds, 95th/99th percentiles, and interquartile range (IQR) outlier boundaries.
- Segment transaction distributions using boolean masking and array slicing.

### Phase 4: Basic Plotting & Volume Trends (Matplotlib)
- Construct frequency histograms to inspect transaction amount distributions.
- Build bar charts comparing aggregate transaction counts across different payment methods.
- Plot temporal line graphs tracking transaction frequencies across hours of the day.
- Customize axis scales, ticks, labels, and figure dimensions for clear visual presentation.

### Phase 5: Categorical Analysis & Statistical Visualizations (Seaborn)
- Generate boxplots to inspect transaction amount distributions across fraud and non-fraud classes.
- Create count plots and categorical bar plots to analyze merchant category distributions.
- Implement kernel density estimate (KDE) and distribution plots for numerical features.
- Apply consistent visual themes, palettes, and styling across all figures.
