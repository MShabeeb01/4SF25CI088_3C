```python
import numpy as np
import pandas as pd

# Load dataset using pandas to extract clean numpy arrays
df = pd.read_csv("indian_banking_transactions.csv")

# Extract transaction_amount as a 1D NumPy array
amounts = df['transaction_amount'].values

print("Data Type :", type(amounts))
print("Array Shape:", amounts.shape)
print("First 5 values:", amounts[:5])
```

```python
# Check for any NaN values
print("Total NaNs in array:", np.isnan(amounts).sum())

# Compute mean ignoring NaNs
mean_amt = np.nanmean(amounts)
print(f"NumPy Mean (ignoring NaNs): {mean_amt:.2f}")

# Replace NaNs with the calculated mean
amounts = np.nan_to_num(amounts, nan=mean_amt)

# Verify no NaNs remain
print("Remaining NaNs:", np.isnan(amounts).sum())
```

```python
# Calculate basic statistical metrics using NumPy built-in functions
mean_val = np.mean(amounts)
median_val = np.median(amounts)
std_val = np.std(amounts)
var_val = np.var(amounts)
min_val = np.min(amounts)
max_val = np.max(amounts)

print("--- Descriptive Statistics ---")
print(f"Mean              : {mean_val:.2f}")
print(f"Median            : {median_val:.2f}")
print(f"Standard Deviation: {std_val:.2f}")
print(f"Variance          : {var_val:.2f}")
print(f"Minimum Value     : {min_val:.2f}")
print(f"Maximum Value     : {max_val:.2f}")
```

```python
# Calculate quartiles and upper percentiles
q1 = np.percentile(amounts, 25)
q2 = np.percentile(amounts, 50)  # Median
q3 = np.percentile(amounts, 75)
p90 = np.percentile(amounts, 90)
p99 = np.percentile(amounts, 99)

print(f"25th Percentile (Q1) : {q1:.2f}")
print(f"50th Percentile (Q2) : {q2:.2f}")
print(f"75th Percentile (Q3) : {q3:.2f}")
print(f"90th Percentile      : {p90:.2f}")
print(f"99th Percentile      : {p99:.2f}")
```

```python
# Interquartile Range (IQR) calculation
iqr = q3 - q1

# Define standard outlier boundaries
lower_limit = q1 - (1.5 * iqr)
upper_limit = q3 + (1.5 * iqr)

# Filter outliers using NumPy boolean masks
outliers = amounts[(amounts < lower_limit) | (amounts > upper_limit)]

print(f"IQR               : {iqr:.2f}")
print(f"Lower Bound       : {lower_limit:.2f}")
print(f"Upper Bound       : {upper_limit:.2f}")
print(f"Total Outliers    : {len(outliers):,}")
print(f"Outlier Percentage: {(len(outliers) / len(amounts)) * 100:.2f}%")
```

```python
# Use boolean indexing to filter transactions above the mean
above_mean = amounts[amounts > mean_val]

print(f"Transactions above average: {len(above_mean):,}")
print(f"Percentage above average  : {(len(above_mean) / len(amounts)) * 100:.2f}%")
```

```python
# Extract a second numeric feature: account_balance
balances = df['account_balance'].values
balances = np.nan_to_num(balances, nan=np.nanmean(balances))

# Stack into a 2D matrix (rows x 2 columns)
matrix_2d = np.column_stack((amounts, balances))

print("2D Matrix Shape:", matrix_2d.shape)
print("Column Means (amounts, balances):", np.mean(matrix_2d, axis=0))

# Compute correlation coefficient matrix between amount and balance
corr_matrix = np.corrcoef(amounts, balances)
print("\nCorrelation Matrix:")
print(np.round(corr_matrix, 4))
```
