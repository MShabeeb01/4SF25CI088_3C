```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("indian_banking_transactions.csv")

# Display the first 5 rows
df.head()
```

```python
# Check total rows and columns
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumns:", df.columns)

# Check data types and non-null values
df.info()
```

```python
# Statistical summary of numerical columns (mean, min, max, etc.)
df.describe()
```

```python
# Count how many null values are present in each column
df.isnull().sum()
```

```python
# 1. Fill categorical nulls in 'loan_type'
df['loan_type'] = df['loan_type'].fillna('No Loan')

# 2. Mean imputation on numerical columns (Milestone 1 requirement)
for col in ['emi_amount', 'transaction_amount', 'account_balance']:
    if col in df.columns:
        mean_val = df[col].mean()
        df[col] = df[col].fillna(mean_val)
        print(f"Mean value used for '{col}': {mean_val:.2f}")

# 3. Verify null counts after filling
df.isnull().sum()
```

```python
# 1. Drop unnecessary columns
if 'transaction_id' in df.columns:
    df = df.drop(columns=['transaction_id'])
    print("Dropped transaction_id column.")

# 2. Add a new derived column
if 'transaction_date' in df.columns:
    df['transaction_date'] = pd.to_datetime(df['transaction_date'])
    df['month'] = df['transaction_date'].dt.month_name()
    print("Added new 'month' column.")

df.head()
```

```python
# Check average transaction amount by channel
summary = df.groupby('channel')['transaction_amount'].mean()
print("Average Transaction Amount by Channel:")
print(summary)
```
